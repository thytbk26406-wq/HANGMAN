```python
import sys
from PyQt6 import QtWidgets
from maxscoregiaodien import Ui_Maxscore
from thuvien_max import diem_cao_nhat, lay_diem_3_level
from chonlevel import Ui_chonlevel

#khởi chạy giao diện
app=QtWidgets.QApplication(sys.argv)
form=Ui_Maxscore()
window=QtWidgets.QWidget()
form.setupUi(window)

# Chưa có trang SELECT LEVEL của nhóm
window_level = QtWidgets.QWidget()
form_level = Ui_chonlevel()
form_level.setupUi(window_level)

# Hàm hiển thị điểm
def hien_thi_diem():
    easy, medium, hard = lay_diem_3_level()

    form.lbl_diemmaxE.setText(str(easy))
    form.lbl_diemmaxM.setText(str(medium))
    form.lbl_diemmaxH.setText(str(hard))

# Hàm quay lại
def quay_lai():
    window_level.show()
    window.hide()

form.btnBack.clicked.connect(quay_lai)

# Hiển thị điểm
hien_thi_diem()

# Mở cửa sổ HIGH SCORES
window.show()
app.exec()
```
