```python
import sys
from PyQt6 import QtWidgets
from chonlevel import Ui_chonlevel
from maxscoregiaodien import Ui_Maxscore
from thuvien_max import lay_diem_3_level
from test import Hangman as HangmanEasy
from test_lv2 import Hangman as HangmanMedium
from test_lv3 import Hangman as HangmanHard

# 1. Khởi tạo ứng dụng
app = QtWidgets.QApplication.instance()
if app is None:
    app = QtWidgets.QApplication(sys.argv)

# 2. Tạo trang SELECT LEVEL
window_level = QtWidgets.QWidget()
form_level = Ui_chonlevel()
form_level.setupUi(window_level)

# 3. Tạo trang HIGH SCORES
window_scores = QtWidgets.QWidget()
form_scores = Ui_Maxscore()
form_scores.setupUi(window_scores)

# 4. Tạo các cửa sổ chơi
window_easy = HangmanEasy()
window_medium = HangmanMedium()
window_hard = HangmanHard()

# 5. Mở level
def mo_level(window_game):
    window_game.show()
    window_level.hide()

# 5. Hàm mở level EASY
def chon_easy():
    mo_level(window_easy)

# 6. Hàm mở level MEDIUM
def chon_medium():
    mo_level(window_medium)

# 7. Hàm mở level HARD
def chon_hard():
    mo_level(window_hard)



# 6. Hiển thị HIGH SCORES
def hien_thi_diem():
    easy, medium, hard = lay_diem_3_level()
    form_scores.lbl_diemmaxE.setText(str(easy))
    form_scores.lbl_diemmaxM.setText(str(medium))
    form_scores.lbl_diemmaxH.setText(str(hard))

# 7. Quay lại SELECT LEVEL
def quay_lai():
    window_scores.hide()
    window_level.show()

# 8. Kết nối các nút
form_level.btnEasy.clicked.connect(chon_easy)
form_level.btnMedium.clicked.connect(chon_medium)
form_level.btnHard.clicked.connect(chon_hard)
form_scores.btnBack.clicked.connect(quay_lai)

# 9. Chạy thử từ HIGH SCORES
hien_thi_diem()
window_scores.show()

app.exec()
```
